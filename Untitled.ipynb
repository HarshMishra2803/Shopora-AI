# Data preprocessing

```python
# 1 - HANDLE THE MISSING DATA
df["Income"] = df["Income"].fillna(df["Income"].median())
```

```python
df.isnull().sum()
```

```python
# FEATURE ENGINNERING
```

```python
df.columns
```

```python
df.head()
```

```python
# Age
df["Age"] = 2026-df["Year_Birth"]
```

```python
df.head()
```

```python
#Customer joining date
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"],dayfirst=True)

refrence_date = df["Dt_Customer"].max()

df["Customer_tenure"] = (refrence_date-df["Dt_Customer"]).dt.days
```

```python
df.head()
```

```python
df.columns
```

```python
# spendings
df["total_spending"] = df["MntFishProducts"]+df["MntMeatProducts"]+df["MntSweetProducts"]+df["MntFruits"]+df["MntWines"]+df["MntGoldProds"]
```

```python
df.head()
```

```python
# childrens

df["total_childrens"] = df["Kidhome"]+df["Teenhome"]
```

```python
df.head()
```

```python
# Education 
df["Education"].value_counts()

df["Education"]=df["Education"].replace(
    {
        "Basic":"UnderGraduate",
        "2n Cycle":" UnderGraduate",
        "Graduation":"Graduate",
        "Master":"PostGraduate",
        "PhD" :"PostGraduate"
    }
)

# UnderGraduate , Graduate , PostGraduate
```

```python
df["Education"].value_counts()
```

```python
# Marital _ Status

df["Marital_Status"].value_counts()

df["Living_with"]=df["Marital_Status"].replace(
    {
        "Married":"Partners",
        "Together":"Partners",
        "Single":"Alone",
        "Divorced":"Alone",
        "Widow" :"Alone",
        "Absurd" :"Alone",
        "YOLO" :"Alone"
    }
)
```

```python
df["Living_with"].value_counts()
```

```python
df.head()
```

```python
# DROP COLUMNS

cols = ["ID", "Year_Birth", "Marital_Status", "Kidhome", "Teenhome", "Dt_Customer"]
spending_cols = ["MntWines", "MntFruits", "MntMeatProducts", "MntFishProducts", "MntSweetProducts", "MntGoldProds"]

cols_to_drop = cols + spending_cols

df_cleaned = df.drop(columns=cols_to_drop)
```

```python
df_cleaned.shape
```

```python
df.shape
```

```python
# Encoding

from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()

cat_cols = ["Education", "Living_with"]

enc_cols = ohe.fit_transform(df_cleaned[cat_cols])
```

```python
enc_df = pd.DataFrame(enc_cols.toarray(), columns=ohe.get_feature_names_out(cat_cols), index=df_cleaned.index)
```

```python
enc_df.head()
```

```python
df_encoded = pd.concat([df_cleaned.drop(columns=cat_cols),enc_df], axis=1)
```

```python
df_encoded.shape
```

```python
# Scaling
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
X = df_encoded
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
df_encoded.head()
```

```python
X_scaled.shape
```

```python
# 2D 
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=3)

X_pca = pca.fit_transform(X_scaled)

pca.explained_variance_ratio_
```

```python
# plot
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2])

ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
ax.set_title("3d projection")
```

```python
# Anlyze k value 
# elbow method


from sklearn.cluster import KMeans
from kneed import KneeLocator

wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
    
```

```python
knee = KneeLocator(range(1, 11), wcss, curve="convex", direction="decreasing")
optimal_k = knee.elbow
```

```python
print("best k =", optimal_k)
```

```python
# plot

plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel("K")
plt.ylabel("WCSS")
```

```python
## 2. Silhouette Score
```

```python
from sklearn.metrics import silhouette_score

scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_pca)
    score = silhouette_score(X_pca, labels)
    scores.append(score)

# plot
plt.plot(range(2, 11), scores, marker='o')
plt.xlabel("K")
plt.ylabel("Silhouette score")
```

```python
# combined plot

k_range = range(2, 11)

fig, ax1 = plt.subplots(figsize=(8, 6))

ax1.plot(k_range, wcss[:len(k_range)], marker="o", color="blue") 
ax1.set_xlabel("K")
ax1.set_ylabel("WCSS")

ax2 = ax1.twinx()
ax2.plot(k_range, scores[:len(k_range)], marker="x", color="red", linestyle="--")
ax2.set_ylabel("SS")
```

```python
# K_means
# CLustering

kmeans = KMeans(n_clusters=3, random_state=42)
labels_kmeans = kmeans.fit_predict(X_pca)
```

```python
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2], c=labels_kmeans)
```

```python
# Agglomerative Clustering
from sklearn.cluster import AgglomerativeClustering
```

```python
agg_clf = AgglomerativeClustering(n_clusters=4, linkage="ward")
labels_agg = agg_clf.fit_predict(X_pca)
```

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")
ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2], c=labels_agg)
```
